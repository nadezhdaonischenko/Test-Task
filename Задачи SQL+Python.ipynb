```python
df = pd.read_sql_query("""
SELECT COUNT(DISTINCT CASE WHEN action='view' THEN user_id END) as view_users,
       COUNT(DISTINCT CASE WHEN action='purchase' THEN user_id END) as purchase_users,
       ROUND(COUNT(DISTINCT CASE WHEN action='purchase' THEN user_id END)*1.0/COUNT(DISTINCT CASE WHEN action='view' THEN user_id END), 2) as CR
FROM events
""", conn)

df
```

Conversion Rate (CR%) = 0.64 (64%)

Получившееся CR великовато, наводит на более глубокий расчет.

```python
df_2 = pd.read_sql_query("""
SELECT
    COUNT(DISTINCT v.user_id) AS converted_users,  -- Пользователи из преобразованной таблицы, которые после просмотра совершили покупку
    (SELECT COUNT(DISTINCT user_id)
     FROM events
     WHERE action='view') AS view_users, -- Все пользователи просматривающие товары
    ROUND(
      COUNT(DISTINCT v.user_id) * 1.0 / (SELECT COUNT(DISTINCT user_id)
                                             FROM events
                                             WHERE action='view'),
    2) AS CR
FROM events v                       -- Пользователи у которых v.action = 'view'
JOIN events p                       -- Только совпадение с пользователями которые в итоге совершили покупку p.action = 'purchase' и время покупки больше времени просмотра
  ON v.user_id = p.user_id
WHERE v.action = 'view'
  AND p.action = 'purchase'
  AND p.event_timestamp > v.event_timestamp
  """, conn)

df_2
```

Conversion Rate (CR%) = 0,5 (50%)

Это конверсия выблядит лучше, более реалистичная, хоть и тоже великовата.

```python
def correct_price(price):
  if price is None:
    return 0

  if '$' in price:
    currency = 'USD'
  elif '€' in price:
    currency = 'EUR'

  price = price.replace('$', '').replace('€', '')

  price = str(price).strip()

  if ',' in price and '.' in price:
    if price.find(',') < price.find('.'):
      price = price.replace(',', '')
    else:
      price = price.replace('.', '').replace(',', '.')

  elif ',' in price:
    price = price.replace(',', '.')

  if price in ['invalid', 'NaN', 'error_val', 'bad_data', 'test', 'broken', 'null', 'N/A']:
    return 0

  corr_price = float(price)

  if currency == 'EUR':
    corr_price = 1.1 * corr_price

  return corr_price


def revenue(transactions):
  total_revenue = 0
  for trans in transactions:
    total_revenue += correct_price(trans['price'])
  return total_revenue

print(f'Сумма выручки по транзакциям равна {round(revenue(transactions),2)} USD')
```
